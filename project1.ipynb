```python
import pandas as pd
```

```python

df = pd.DataFrame({
    "name": ["Ram", "Shyam", "Hari"],
    "math": [99, 76, 67],
    "science": [89, 65, 76],
    "english": [65, 43, 65]
})
```

```python
df["passed"] = (
    (df["math"] >= 60) &
    (df["science"] >= 60) &
    (df["english"] >= 60)
)

print(df)
```

```python
fail_student = df[~df["passed"]]
fail_student
```

```python
result = []

for value in df["passed"]:
    if value:
        result.append("pass")
    else:
        result.append("fail")
df["Result"] = result
df
```

```python
fail_student = df[df["Result"] == "fail"]
fail_student
```

```python
count = 0
for values in  df["Result"]:
    if values =="pass":
       count+=1
print("passing student:",count)
```

```python
count = 0
for values in  df["Result"]:
    if values =="fail":
       count+=1
print("failing student:",count)
```

```python
df["total"] = df["math"] + df["science"] + df["english"]
df
```

```python
highest = -1
topper = ""
for i in range(len(df)):
    if df.loc[i,"total"]>highest:
        highest = df.loc[i,"total"]
        topper = df.loc[i,"name"]
print("topper:",topper)
print("Marks:",highest)
```

```python
subjects = ["math", "science", "english"]

df["average"] = df[subjects].mean(axis=1)
print(df)
```

```python
subjects = ["math", "science", "english"]
average = []
for i in range(len(df)):
    total = 0
    for subject in subjects:
        total += df.loc[i,subject]
    average.append(total/len(subjects))

df["average"]=average
print(df)
    
```

```python
grades=[]
for average in df["average"]:
    if average>=80:
        grades.append("A")
    elif average>=60:
        grades.append("B")
    else:
        grades.append("C")
df["Grade"]=grades
print(df)
```

```python
grades = []

for i in range(len(df)):
    if (df.loc[i, "math"] < 60 or
        df.loc[i, "science"] < 60 or
        df.loc[i, "english"] < 60):
       
        grades.append("NG")

    elif df.loc[i, "average"] >= 80:
        grades.append("A+")

    elif df.loc[i, "average"] >= 60:
        grades.append("B+")

    else:
        grades.append("Fail")

df["Grade"] = grades

print(df)
```
