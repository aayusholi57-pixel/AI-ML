```python
import numpy as np
import pandas as pd

from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
df = pd.read_csv(r"C:\Users\AB\Downloads\student_multiple_features.csv")
```

```python
print(df.isnull().sum())
```

```python
df["PreviousMarks"] = df["PreviousMarks"].fillna(
    df["PreviousMarks"].median()
)
df["Attendance"] = df["Attendance"].replace("Absent", 0)
```

```python
df
```

```python
# Features (X)
X = df[
    [
        "StudyHours",
        "Attendance",
        "PreviousMarks"
    ]
]

# Target (Y)
Y = df["FinalMarks"]

# Convert to NumPy arrays
X = X.to_numpy()
Y = Y.to_numpy()

# Split data into training and testing sets
X_tr, X_te, Y_tr, Y_te = train_test_split(
    X,
    Y,
    test_size=0.2,
    random_state=42
)

# Create Linear Regression model
model = LinearRegression()

# Train the model
model.fit(X_tr, Y_tr)

# Make predictions
Y_pred = model.predict(X_te)

# Calculate R² score
r2 = r2_score(Y_te, Y_pred)

# Display results
print( r2)
```
