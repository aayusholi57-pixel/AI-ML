```python
from sklearn.linear_model import LinearRegression
import numpy as np
from sklearn.metrics import mean_squared_error, r2_score
```

```python
model = LinearRegression()
```

```python
x_train = np.array([
    [600],
    [750],
    [900],
    [1050],
    [1200],
    [1350],
    [1500],
    [1650],
    [1800],
    [1950],
    [2100],
    [2250],
    [2400],
    [2550],
    [2700]
])

# House price in thousands of dollars (Y)
y_train = np.array([
    120,
    145,
    170,
    190,
    215,
    240,
    265,
    285,
    310,
    335,
    360,
    385,
    405,
    430,
    455
])
```

```python
y_test = np.array ([
    285,
    310,
    335,
    360,
    385,
    405,
    430,
    455
])
```

```python
x_test = np.array ([
     [1650],
    [1800],
    [1950],
    [2100],
    [2250],
    [2400],
    [2550],
    [2700]
])
```

```python
model.fit(x_train, y_train)
```

```python
model.predict([[850],[659]])
```

```python
y_preds = model.predict(x_test)
```

```python
y_preds
```

```python
rmse =np.sqrt(
    mean_squared_error(y_test,y_preds)
)
```
