```python
import pandas as pd
import random

names = [
    "Ram", "Sita", "Hari", "Gita", "Shyam",
    "Rita", "Aayush", "Kiran", "Bikash", "Anita"
]

data = {
    "Name": [],
    "Age": [],
    "Hours_Studied": [],
    "Marks": []
}

for i in range(100):
    data["Name"].append(random.choice(names))
    data["Age"].append(random.randint(18, 24))
    hours = random.randint(1, 8)
    data["Hours_Studied"].append(hours)
    data["Marks"].append(random.randint(40 + hours * 5, 100))

df = pd.DataFrame(data)

```

```python
df.to_csv("task.csv", index=False)

print("Dataset Created!")
```
