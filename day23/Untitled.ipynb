```python
from review_data import make_reviews

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline
from sklearn.model_selection import cross_val_score


# Get our reviews
texts, labels = make_reviews()


# Create the AI pipeline
model = make_pipeline(
    TfidfVectorizer(),
    LogisticRegression()
)


# Test the model using 5-fold cross-validation
scores = cross_val_score(
    model,
    texts,
    labels,
    cv=5
)


print("Scores:", scores)
print("Average:", round(scores.mean(), 3))
```

```python
# Train the model on all our reviews
model.fit(texts, labels)


# New reviews the AI has never seen
tests = [
    "The food was wonderful",
    "The service was awful",
    "I really enjoyed this restaurant",
    "The meal was horrible",
]


# Make predictions
predictions = model.predict(tests)


# Show the results
for text, prediction in zip(tests, predictions):

    if prediction == 1:
        result = "POSITIVE"
    else:
        result = "NEGATIVE"

    print(result, "<-", text)
```

```python
tests = [
    "the food was not good",
    "not bad at all",
    "I would not say the service was terrible"
]

for t, p in zip(tests, model.predict(tests)):

    print(
        "POSITIVE" if p == 1 else "NEGATIVE",
        "<-",
        t
    )
```
