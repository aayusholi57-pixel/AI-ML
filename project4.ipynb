```python
class Book:

    def __init__(self, book_id, title, author, price, quantity):
        self.book_id = book_id
        self.title = title
        self.author = author
        self.price = price
        self.quantity = quantity
```

```python
book1 = Book(101, "Python", "John", 3000, 10)
book2 = Book(102, "AI", "David", 2000, 9)
book3 = Book(103, "Numpy", "Simon", 1000, 8)
book4 = Book(104, "Machine Learning", "Michael", 999, 7)
```

```python
class Library:

    def __init__(self):
        self.books = []

    def add_book(self, book):
        self.books.append(book)
    def display_books(self):
        for book in self.books:
           print("......................")
           print("Book ID :", book.book_id)
           print("Title   :", book.title)
           print("Author  :", book.author)
           print("Price   :", book.price)
           print("Quantity:", book.quantity)
library = Library()
```

```python
library.add_book(book1)
library.add_book(book2)
library.add_book(book3)
library.add_book(book4)
```

```python
library.display_books()
```
