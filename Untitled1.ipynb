```python
import matplotlib.pyplot as plt
import numpy as np 
import pandas as pd
import seaborn as sns
import plotly.express as ex
import matplotlib.patheffects as path_effects

students_data={
    'junu':85,
    'keshab':91,
    'krishna':12,
    'charlie':15,
    'David':20,
    'Ella':40
}
df=pd.DataFrame(list(students_data.items()),columns=['Name','Score'])
df=df.sort_values(by='Score',ascending=False)

sns.set_theme(style='whitegrid')
palette=sns.color_palette("Spectral",len(df))

#we will add the plots
plt.figure(figsize=(30,20))
barplot=sns.barplot(data=df,x='Name',y='Score',palette=palette,edgecolor='black')

for bar in barplot.patches:
    x=bar.get_x()+bar.get_width()/2
    y=bar.get_height()
    label=plt.text(x,y+1,f'{int(y)}',ha='center',va='bottom',fontsize=40
                   ,fontweight='bold',color='#2cceee')
    label.set_path_effects([path_effects.Stroke(linewidth=1.5,foreground='white'),path_effects.Normal()])

plt.title('student performace',fontsize=14,fontweight='bold',color='#2c2c2c',pad=20)
plt.xlabel('Student Name',fontsize=12)
plt.ylabel('Score',fontsize=20)
plt.xticks(rotation=90,fontsize=20)
plt.yticks(fontsize=20)


ax=plt.gca()
ax.set_facecolor('#fafafa')
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.grid(axis='y',linestyle='--',alpha=0.7)

plt.tight_layout(pad=0.2)
plt.show()
```
