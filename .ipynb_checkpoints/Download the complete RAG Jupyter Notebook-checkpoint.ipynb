# Complete RAG Using TF-IDF, TruncatedSVD, Normalizer, and Pipelines

This notebook builds a complete educational **Retrieval-Augmented Generation (RAG)** system.

## Retrieval flow

```text
Documents → Chunking → TF-IDF → TruncatedSVD → L2 Normalizer → Document vectors
Question  → Same fitted pipeline                            → Query vector
Document vectors + Query vector → Cosine similarity → Top-k chunks
Top-k chunks + Question → Hugging Face generation pipeline → Grounded answer
```

## Components used

- `TfidfVectorizer`: converts text into sparse term-weight vectors.
- `TruncatedSVD`: reduces dimensionality and learns latent semantic patterns.
- `Normalizer`: scales each vector to unit L2 length.
- `sklearn.pipeline.Pipeline`: chains the retrieval transformations.
- `transformers.pipeline`: runs the answer-generation model.

> The first model-generation run downloads a Hugging Face model, so an internet connection is needed once. Retrieval itself runs locally.

## 2. Import libraries

The two different pipeline classes are given aliases to avoid a naming conflict.


````python
from dataclasses import dataclass
from pathlib import Path
from typing import Any, Dict, List, Optional
import warnings

import joblib
import numpy as np

from sklearn.decomposition import TruncatedSVD
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.pipeline import Pipeline as SklearnPipeline
from sklearn.preprocessing import Normalizer

from transformers import pipeline as hf_pipeline

warnings.filterwarnings("ignore")
np.set_printoptions(precision=4, suppress=True)

print("Imports completed successfully.")
````

## 3. Define documents and chunks

Metadata is retained so retrieved passages can be cited by source and chunk number.


````python
@dataclass
class Document:
    text: str
    metadata: Dict[str, Any]


@dataclass
class Chunk:
    text: str
    metadata: Dict[str, Any]
````

## 4. Create the SVD retriever

Important details:

1. Documents are split into overlapping word-based chunks.
2. TF-IDF learns word and bigram importance.
3. TruncatedSVD maps sparse vectors into a smaller dense latent space.
4. L2 normalization makes the dot product equivalent to cosine similarity.
5. The query uses `transform()`, not `fit_transform()`, so it stays in the same vector space as the documents.


````python
class SVDDocumentRetriever:
    """Retrieve text chunks using TF-IDF, TruncatedSVD, and L2 normalization."""

    def __init__(
        self,
        n_components: int = 100,
        chunk_size: int = 80,
        chunk_overlap: int = 20,
        stop_words: Optional[str] = "english",
    ):
        if chunk_size <= 0:
            raise ValueError("chunk_size must be greater than zero.")
        if chunk_overlap < 0:
            raise ValueError("chunk_overlap cannot be negative.")
        if chunk_overlap >= chunk_size:
            raise ValueError("chunk_overlap must be smaller than chunk_size.")
        if n_components <= 0:
            raise ValueError("n_components must be greater than zero.")

        self.n_components = n_components
        self.chunk_size = chunk_size
        self.chunk_overlap = chunk_overlap
        self.stop_words = stop_words

        self.chunks: List[Chunk] = []
        self.document_vectors: Optional[np.ndarray] = None
        self.retrieval_pipeline: Optional[SklearnPipeline] = None

    def chunk_document(self, document: Document) -> List[Chunk]:
        words = document.text.split()
        chunks: List[Chunk] = []

        if not words:
            return chunks

        step = self.chunk_size - self.chunk_overlap

        for chunk_number, start in enumerate(range(0, len(words), step), start=1):
            end = min(start + self.chunk_size, len(words))
            chunk_words = words[start:end]

            if not chunk_words:
                continue

            metadata = {
                **document.metadata,
                "chunk_number": chunk_number,
                "start_word": start,
                "end_word": end,
            }
            chunks.append(Chunk(text=" ".join(chunk_words), metadata=metadata))

            if end == len(words):
                break

        return chunks

    def create_chunks(self, documents: List[Document]) -> List[Chunk]:
        if not documents:
            raise ValueError("At least one document is required.")

        self.chunks = []
        for document in documents:
            self.chunks.extend(self.chunk_document(document))

        if len(self.chunks) < 2:
            raise ValueError(
                "At least two non-empty chunks are required for TruncatedSVD. "
                "Add documents or reduce chunk_size."
            )
        return self.chunks

    def build_index(self, documents: List[Document]) -> None:
        self.create_chunks(documents)
        texts = [chunk.text for chunk in self.chunks]

        # SVD dimensions must be valid for the fitted TF-IDF matrix.
        # The vocabulary size is not known until TF-IDF is fitted, so fit it first.
        tfidf = TfidfVectorizer(
            lowercase=True,
            stop_words=self.stop_words,
            ngram_range=(1, 2),
            min_df=1,
            sublinear_tf=True,
        )
        tfidf_matrix = tfidf.fit_transform(texts)

        max_valid_components = min(tfidf_matrix.shape[0] - 1, tfidf_matrix.shape[1] - 1)
        safe_components = min(self.n_components, max_valid_components)
        safe_components = max(1, safe_components)

        self.retrieval_pipeline = SklearnPipeline(
            steps=[
                ("tfidf", tfidf),
                ("svd", TruncatedSVD(n_components=safe_components, random_state=42)),
                ("normalizer", Normalizer(norm="l2", copy=False)),
            ]
        )

        # TF-IDF is already fitted. Calling fit_transform on the full pipeline
        # refits TF-IDF consistently and then fits SVD and Normalizer.
        self.document_vectors = self.retrieval_pipeline.fit_transform(texts)

        print("Index created successfully.")
        print(f"Chunks: {len(self.chunks)}")
        print(f"TF-IDF vocabulary size: {len(self.retrieval_pipeline.named_steps['tfidf'].vocabulary_)}")
        print(f"SVD components: {safe_components}")
        print(f"Document-vector shape: {self.document_vectors.shape}")

    def retrieve(self, question: str, top_k: int = 3) -> List[Dict[str, Any]]:
        if self.retrieval_pipeline is None or self.document_vectors is None:
            raise RuntimeError("Build or load the index before retrieving.")
        if not question.strip():
            raise ValueError("The question cannot be empty.")
        if top_k <= 0:
            raise ValueError("top_k must be greater than zero.")

        query_vector = self.retrieval_pipeline.transform([question])
        scores = np.asarray(self.document_vectors @ query_vector.T).ravel()
        top_indices = np.argsort(scores)[::-1][: min(top_k, len(self.chunks))]

        return [
            {
                "rank": rank,
                "score": float(scores[index]),
                "text": self.chunks[index].text,
                "metadata": self.chunks[index].metadata,
            }
            for rank, index in enumerate(top_indices, start=1)
        ]

    def save(self, path: str = "svd_retriever.joblib") -> None:
        if self.retrieval_pipeline is None or self.document_vectors is None:
            raise RuntimeError("There is no built index to save.")
        joblib.dump(self, path)
        print(f"Retriever saved to: {Path(path).resolve()}")

    @staticmethod
    def load(path: str = "svd_retriever.joblib") -> "SVDDocumentRetriever":
        retriever = joblib.load(path)
        print(f"Retriever loaded from: {Path(path).resolve()}")
        return retriever
````

## 5. Prepare a sample knowledge base

````python
documents = [
    Document(
        text="""
        Retrieval-Augmented Generation, commonly called RAG, combines information
        retrieval with language generation. When a user asks a question, a retriever
        searches an external knowledge collection for relevant passages. The passages
        are inserted into the language model prompt. The model then generates an answer
        from the retrieved evidence. RAG can use private or recently updated knowledge
        without retraining the complete language model. It can reduce unsupported claims,
        but it does not guarantee that every generated answer will be correct.
        """,
        metadata={"source": "rag_introduction.txt", "topic": "RAG"},
    ),
    Document(
        text="""
        TF-IDF means term frequency-inverse document frequency. Term frequency measures
        how often a term appears in a document. Inverse document frequency lowers the
        weight of terms that occur in many documents. TF-IDF produces sparse vectors and
        is commonly used for document retrieval, classification, and keyword extraction.
        Words and phrases that are important to a particular document receive larger weights.
        """,
        metadata={"source": "tfidf_notes.txt", "topic": "TF-IDF"},
    ),
    Document(
        text="""
        TruncatedSVD is a dimensionality-reduction algorithm that works efficiently with
        sparse matrices such as TF-IDF matrices. When it is applied to term-document data,
        it is associated with latent semantic analysis. It creates smaller dense vectors
        that represent important latent patterns. Dimensionality reduction can decrease
        storage and computation while capturing relationships beyond exact term matching.
        """,
        metadata={"source": "svd_notes.txt", "topic": "SVD"},
    ),
    Document(
        text="""
        A normalizer scales each sample independently to unit norm. L2 normalization divides
        a vector by its Euclidean length. When document vectors and query vectors are both
        L2 normalized, their dot product equals cosine similarity. Semantically or lexically
        related vectors should receive larger similarity scores than unrelated vectors.
        """,
        metadata={"source": "normalization_notes.txt", "topic": "Normalization"},
    ),
    Document(
        text="""
        A machine learning pipeline chains transformations in a fixed sequence. During
        training, the required components are fitted in order. During inference, the same
        fitted transformations are applied to new input. A retrieval pipeline can connect
        TF-IDF vectorization, TruncatedSVD dimensionality reduction, and vector normalization.
        This prevents accidental differences between document and query preprocessing.
        """,
        metadata={"source": "pipeline_notes.txt", "topic": "Pipeline"},
    ),
    Document(
        text="""
        RAG evaluation should examine retrieval and generation separately. Retrieval metrics
        include recall at k, precision at k, and mean reciprocal rank. Generation can be
        evaluated for correctness, relevance, completeness, faithfulness to retrieved context,
        and citation quality. If an answer is wrong, first check whether the correct evidence
        was retrieved. If it was retrieved, then investigate how the generator used it.
        """,
        metadata={"source": "rag_evaluation.txt", "topic": "Evaluation"},
    ),
]

print(f"Loaded {len(documents)} documents.")
````

## 6. Build the retrieval index

````python
retriever = SVDDocumentRetriever(
    n_components=50,
    chunk_size=55,
    chunk_overlap=12,
    stop_words="english",
)

retriever.build_index(documents)
````

## 7. Test retrieval before adding the generator

Always evaluate retrieval independently. If the relevant passage is not retrieved, the generator cannot reliably answer from it.


````python
question = "Why do we normalize vectors in this RAG system?"
results = retriever.retrieve(question, top_k=3)

print("QUESTION:", question)
print()

for result in results:
    print(f"Rank: {result['rank']}")
    print(f"Score: {result['score']:.4f}")
    print(f"Source: {result['metadata']['source']}")
    print(f"Text: {result['text']}")
    print("-" * 90)
````

## 8. Optional retrieval-only answer

This function displays retrieved evidence even if a generation model is unavailable.


````python
def show_retrieval(question: str, top_k: int = 3) -> None:
    results = retriever.retrieve(question, top_k=top_k)
    print(f"Question: {question}\n")
    for item in results:
        source = item["metadata"].get("source", "Unknown source")
        print(f"[{item['rank']}] {source} | score={item['score']:.4f}")
        print(item["text"])
        print()


show_retrieval("How does TruncatedSVD help TF-IDF retrieval?", top_k=3)
````

## 9. Create the complete RAG class

The default generator is `google/flan-t5-small`, which is convenient for learning on CPU. Larger instruction-tuned models may produce better answers but need more memory and computation.


````python
class SVDRAG:
    """Complete RAG system with an SVD retriever and Hugging Face generator."""

    def __init__(
        self,
        retriever: SVDDocumentRetriever,
        generator_model: str = "google/flan-t5-small",
        device: int = -1,
    ):
        self.retriever = retriever
        print(f"Loading generator: {generator_model}")
        self.generator = hf_pipeline(
            task="text2text-generation",
            model=generator_model,
            tokenizer=generator_model,
            device=device,
        )
        print("Generator loaded successfully.")

    @staticmethod
    def build_prompt(question: str, results: List[Dict[str, Any]]) -> str:
        sections = []
        for item in results:
            source = item["metadata"].get("source", "Unknown source")
            chunk_number = item["metadata"].get("chunk_number", "?")
            sections.append(
                f"[Source: {source}, chunk: {chunk_number}]\n{item['text']}"
            )

        context = "\n\n".join(sections)

        return f"""Answer the question using only the context below.
If the context does not contain enough information, answer exactly:
I do not have enough information in the provided documents.
Do not add unsupported facts. Give a concise answer and mention the source name.

Context:
{context}

Question: {question}
Answer:"""

    def ask(
        self,
        question: str,
        top_k: int = 3,
        max_new_tokens: int = 120,
    ) -> Dict[str, Any]:
        results = self.retriever.retrieve(question, top_k=top_k)
        prompt = self.build_prompt(question, results)

        output = self.generator(
            prompt,
            max_new_tokens=max_new_tokens,
            do_sample=False,
            truncation=True,
        )
        answer = output[0]["generated_text"].strip()

        sources = [
            {
                "source": item["metadata"].get("source", "Unknown source"),
                "chunk_number": item["metadata"].get("chunk_number"),
                "similarity_score": round(item["score"], 4),
                "retrieved_text": item["text"],
            }
            for item in results
        ]

        return {
            "question": question,
            "answer": answer,
            "sources": sources,
            "prompt": prompt,
        }

    @staticmethod
    def display(result: Dict[str, Any]) -> None:
        print("QUESTION")
        print(result["question"])
        print("\nANSWER")
        print(result["answer"])
        print("\nRETRIEVED SOURCES")
        for source in result["sources"]:
            print(
                f"- {source['source']} | chunk {source['chunk_number']} | "
                f"score {source['similarity_score']:.4f}"
            )
````

## 10. Load the generation model

If you have a CUDA GPU, change `device=-1` to `device=0`.


````python
rag = SVDRAG(
    retriever=retriever,
    generator_model="google/flan-t5-small",
    device=-1,  # CPU. Use 0 for the first CUDA GPU.
)
````

````python
result = rag.ask(
    question="Why is TruncatedSVD useful after TF-IDF?",
    top_k=3,
    max_new_tokens=120,
)

rag.display(result)
````

## 16. Add your own documents

Use the following pattern, then rebuild the index and create a new `SVDRAG` object.

```python
my_documents = [
    Document(
        text="Your first document text...",
        metadata={"source": "document_1.txt", "topic": "Your topic"},
    ),
    Document(
        text="Your second document text...",
        metadata={"source": "document_2.txt", "topic": "Your topic"},
    ),
]

my_retriever = SVDDocumentRetriever(
    n_components=100,
    chunk_size=100,
    chunk_overlap=20,
)
my_retriever.build_index(my_documents)

my_rag = SVDRAG(my_retriever, generator_model="google/flan-t5-small")
my_rag.display(my_rag.ask("Your question here"))
```

## Notes

- Use `fit_transform()` for the document collection.
- Use only `transform()` for new questions.
- Increase `n_components` for a larger document collection, but keep it below the matrix limits.
- Tune `chunk_size`, `chunk_overlap`, and `top_k` using evaluation questions.
- A modern embedding model will usually capture semantics better than TF-IDF plus SVD, but this implementation clearly demonstrates the complete RAG mechanism.
- For factual systems, inspect retrieved passages and do not rely only on fluent generated text.
