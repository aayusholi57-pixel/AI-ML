```python
import pandas as pd
```

```python
student = pd.DataFrame({
    "id": [1, 2, 3, 4, 5],
    "name":["ram","sita","hari","gita","som"],
    "subject": ["math", "science", "health", "AI", "ML"]
})

score = pd.DataFrame({
    "id": [1, 2, 3, 4, 5],
    "city": ["Kathmandu", "Pokhara", "Kohalpur", "Nepalgunj", "Kathmandu"],
    "score": [77, 65, 45, 54, 76]
})
```

```python
student.to_csv("students.csv", index=False)
score.to_csv("scores.csv", index = False)
```

```python
student
```

```python
score
```

```python
merged_df = pd.merge(student, score, on="id")
```

```python
merged_df
```

```python
merged_df.groupby("city")
```

```python
merged_df.groupby("subject")["score"].mean()
```

```python
merged_df.groupby("city")["score"].mean()
```

```python
high_scores = merged_df[merged_df["score"] == merged_df["score"].max()]

high_scores
```

```python
summary = merged_df.groupby("city").agg({
    "score": "mean",
    "name": "count"
}).reset_index()

summary
```

```python
merged_df.sort_values(by="score", ascending=False)
```
