```python
fd["attendance_percent"]=fd["attendance_percent"].fillna(fd["attendance_percent"].median())#fillingmissingvalues
fd.isnull().sum()
```

```python
fd.dtypes
```

```python
fd["enrollment_date"].dtype#converting incorrect data types to appropriate formats
fd["enrollment_date"]=pd.to_datetime(fd["enrollment_date"])
fd.info()
```

```python
fd.columns
```

```python
fd["city"].unique()
fd["city"]=fd["city"].replace("ktm","Kathmandu")
fd["city"].unique()
```

```python
fd["course"].unique()
fd["course"]=fd["course"].replace("Pythn AI", "Python with AI/ML")
fd["course"].unique()
```

```python
fd["gender"].unique()
```

```python
fd["mentor"].unique()
```

```python
fd["completion_status"].unique()
```

```python
fd[fd["course_fee"] < 0]
fd["course_fee"] = fd["course_fee"].replace(-5000, fd["course_fee"].median())
fd[fd["course_fee"] < 0]
```

```python
fd[fd["assignment_score"] < 0]
fd["assignment_score"]=fd["assignment_score"].replace(-5,fd["assignment_score"].median())
fd[fd["assignment_score"] < 0]
```

```python
#outliersfor course_fee
q1=fd["course_fee"].quantile(0.25)
q3=fd["course_fee"].quantile(0.75)
iqr=q3-q1
lower_bound=q1-1.5*iqr
upper_bound=q3+1.5*iqr
print("q1",q1)
print("q3",q3)
print("iqr",iqr)
print("lower_bound",lower_bound)
print("upper_bound",upper_bound)
outliers=fd[(fd["course_fee"]<lower_bound) | (fd["course_fee"] > upper_bound)]
print("outliers",outliers)
fd["course_fee"] = pd.to_numeric(
    fd["course_fee"],
    errors="coerce"
)
fd_clean = fd[
    (fd["course_fee"] >= lower_bound) &
    (fd["course_fee"] <= upper_bound)
]
```

```python
#outlier for assignmentscore
q1=fd["attendance_percent"].quantile(0.25)
q3=fd["attendance_percent"].quantile(0.75)

iqr=q3-q1
lower_bound=q1-1.5*iqr
upper_bound=q3+1.5*iqr
print("q1",q1)
print("q3",q3)
print("iqr",iqr)
print("lower_bound",lower_bound)
print("upper_bound",upper_bound)
outliers=fd[(fd["attendance_percent"]<lower_bound) | (fd["attendance_percent"] > upper_bound)]
print("outliers",outliers)
```
