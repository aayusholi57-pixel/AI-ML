```python
import numpy as np
import fitz


from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD
from sklearn.preprocessing import Normalizer
from sklearn.pipeline import make_pipeline
```

```python
doc = fitz.open("AI Engineering & Machine Learning - Syllabus.pdf")
```

```python
text = ""

for page in doc:
    text += page.get_text() + "\n"

chunk_size =300
overlap = 30

chunks = []

start = 0

while start < len(text):
    end = start + chunk_size

    chunks.append(text[start:end])

    start += chunk_size - overlap

print("Total chunks:", len(chunks))

for i, chunk in enumerate(chunks, start=1):

    print(f"Chunk {i}")
    print(chunk)
```

```python
embedder = make_pipeline(
    TfidfVectorizer(
        stop_words="english",
        max_features=10000
    ),
    TruncatedSVD(
        n_components=20,
        random_state=0
    ),
    Normalizer()
)

index = embedder.fit_transform(chunks)

def retrieve(question, k=1):
    q = embedder.transform([question])[0]
    scores = index @ q
    return [
        (scores[i], chunks[i])
        for i in np.argsort(-scores)[:k]
    ]
```

```python
retrieve("what is linear regression")
```
