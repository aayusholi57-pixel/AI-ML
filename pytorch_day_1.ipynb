### Autograd

```python
x = torch.tensor(3.0, requires_grad=True)
y = x ** 2          # record the operation
y.backward()          # walk the graph backwards
print(x.grad) 
```

```python
x_in   = torch.tensor(2.0)                    # the input
target = torch.tensor(3.0)                    # what we wanted
w = torch.tensor(0.5, requires_grad=True)     # weight to learn
b = torch.tensor(0.0, requires_grad=True)     # bias to learn

z = w * x_in + b                            # forward pass
loss = (z - target) ** 2
loss.backward()                              # fill the gradients
print(z.item(), loss.item())   # 1.0   4.0
print(w.grad, b.grad)    
```

```python
X = torch.linspace(-1, 1, 50);  Y = 2 * X + 1
w = torch.tensor(0.0, requires_grad=True); b = torch.tensor(0.0, requires_grad=True)

for step in range(200):
    
    pred = w * X + b                     # 1. forward
    loss = ((pred - Y) ** 2).mean()     # 2. how wrong (MSE)
    print(f"step {step} the loss is: {loss}")
    loss.backward()                      # 3. autograd fills w.grad, b.grad
    with torch.no_grad():               # 4. step: nudge, don't track this
        w -= 0.1 * w.grad;  b -= 0.1 * b.grad
        w.grad.zero_(); b.grad.zero_()
```
