```python
import numpy as np


# Our handmade word vectors
word_vectors = {

    # [royalty, gender, human, food]

    "king":   np.array([1, 1, 1, 0]),
    "queen":  np.array([1, 0, 1, 0]),

    "man":    np.array([0, 1, 1, 0]),
    "woman":  np.array([0, 0, 1, 0]),

    # New words
    "momo":   np.array([0, 0, 0, 1]),
    "curry":  np.array([0, 0, 0, 1]),
    "prince": np.array([1, 1, 1, 0]),
}
word_vectors
```

```python
# ------------------------------------------------
# 1. Look at our vectors
# ------------------------------------------------

for word, vector in word_vectors.items():
    print(word, "->", vector)
```

```python
# ------------------------------------------------
# 2. Calculate distance between two words
# ------------------------------------------------

def distance(word1, word2):

    v1 = word_vectors[word1]
    v2 = word_vectors[word2]

    return np.linalg.norm(v1 - v2)


print("\nDistance between momo and curry:")
print(distance("momo", "curry"))

print("\nDistance between momo and king:")
print(distance("momo", "king"))
```

```python
# ------------------------------------------------
# 3. King - man + woman
# ------------------------------------------------

result = (
    word_vectors["king"]
    - word_vectors["man"]
    + word_vectors["woman"]
)

print("\nking - man + woman =", result)

```

```python
# 4. Find the closest word to the result
# ------------------------------------------------

best_word = None
best_distance = float("inf")

for word, vector in word_vectors.items():

    d = np.linalg.norm(result - vector)

    if d < best_distance:
        best_distance = d
        best_word = word


print("Closest word:", best_word)
```
