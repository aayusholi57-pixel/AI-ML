```python
# ==========================================
# NLP TEXT CLASSIFICATION
# CountVectorizer + TF-IDF + Pipeline
# ==========================================

import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
















```

```python
# ==========================================
# 1. DATASET
# ==========================================

documents = [
    "I love machine learning",
    "I love Python programming",
    "Python is great for machine learning",
    "I hate boring programming",
    "This movie is amazing",
    "This movie is terrible",
    "I really enjoyed this movie",
    "I dislike this boring movie"
]

labels = [
    "positive",
    "positive",
    "positive",
    "negative",
    "positive",
    "negative",
    "positive",
    "negative"
]
```

```python
# ==========================================
# 2. CREATE DATAFRAME
# ==========================================

df = pd.DataFrame({
    "text": documents,
    "label": labels
})

print("Dataset:")
print(df)

```

```python

# ==========================================
# 3. TRAIN / TEST SPLIT
# ==========================================

X = df["text"]
y = df["label"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

```

```python
# ==========================================
# 4. COUNT VECTORIZER PIPELINE
# ==========================================

count_pipeline = Pipeline([
    ("vectorizer", CountVectorizer(
        lowercase=True,
        stop_words="english",
        ngram_range=(1, 2)
    )),

    ("classifier", LogisticRegression(
        max_iter=1000,
        random_state=42
    ))
])
```

```python
# ==========================================
# 5. TRAIN COUNT VECTORIZER MODEL
# ==========================================

count_pipeline.fit(X_train, y_train)
```

```python
# ==========================================
# 6. COUNT VECTORIZER PREDICTION
# ==========================================

count_predictions = count_pipeline.predict(X_test)

print("\n========== COUNT VECTORIZER ==========")

print("Accuracy:",
      accuracy_score(y_test, count_predictions))

print("\nClassification Report:")
print(classification_report(y_test, count_predictions))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, count_predictions))

```

```python
# ==========================================
# 7. TF-IDF PIPELINE
# ==========================================

tfidf_pipeline = Pipeline([
    ("vectorizer", TfidfVectorizer(
        lowercase=True,
        stop_words="english",
        ngram_range=(1, 2),
        min_df=1,
        max_df=0.95,
        sublinear_tf=True
    )),

    ("classifier", LogisticRegression(
        max_iter=1000,
        random_state=42
    ))
])

```

```python
# ==========================================
# 8. TRAIN TF-IDF MODEL
# ==========================================

tfidf_pipeline.fit(X_train, y_train)
```

```python
# ==========================================
# 9. TF-IDF PREDICTION
# ==========================================

tfidf_predictions = tfidf_pipeline.predict(X_test)

print("\n========== TF-IDF ==========")

print("Accuracy:",
      accuracy_score(y_test, tfidf_predictions))

print("\nClassification Report:")
print(classification_report(y_test, tfidf_predictions))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, tfidf_predictions))
```

```python
# ==========================================
# 10. TEST WITH NEW TEXT
# ==========================================

new_text = [
    "I really love this amazing movie",
    "This is a terrible and boring movie"
]

print("\n========== NEW PREDICTIONS ==========")

print("CountVectorizer:")
print(count_pipeline.predict(new_text))

print("\nTF-IDF:")
print(tfidf_pipeline.predict(new_text))
```
