```python
import torch
import torch.nn as nn

X = torch.tensor([
    [1.0],
    [2.0],
    [3.0],
    [4.0]
])

Y = torch.tensor([
    [3.0],
    [5.0],
    [7.0],
    [9.0]
])

model = nn.Linear(1, 1)

loss_fn = nn.MSELoss()

optimizer = torch.optim.SGD(
    model.parameters(),
    lr=0.1
)

for step in range(200):

    pred = model(X)

    loss = loss_fn(pred, Y)

    optimizer.zero_grad()

    loss.backward()

    optimizer.step()

print("weight:", model.weight.item())
print("bias:", model.bias.item())
```
