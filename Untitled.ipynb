```python
import numpy as np

from sklearn.neural_network import MLPClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import cross_val_score

# ------------------------------------
# Build Dataset
# Healthy if inside the circle:
# x² + y² < 4
# ------------------------------------

rng = np.random.default_rng(42)
n = 600

x = rng.uniform(-3, 3, n)
y = rng.uniform(-3, 3, n)

healthy = ((x**2 + y**2) < 4).astype(int)

X = np.column_stack([x, y])

# ------------------------------------
# Logistic Regression Model
# ------------------------------------

logistic = make_pipeline(
    StandardScaler(),
    LogisticRegression(max_iter=1000)
)

# ------------------------------------
# Neural Network Model
# ------------------------------------

net = make_pipeline(
    StandardScaler(),
    MLPClassifier(
        hidden_layer_sizes=(16, 16),
        max_iter=2000,
        random_state=42
    )
)

# ------------------------------------
# Evaluate Models
# ------------------------------------

logistic_score = cross_val_score(logistic, X, healthy, cv=5).mean()
net_score = cross_val_score(net, X, healthy, cv=5).mean()

print("Logistic Regression Accuracy :", logistic_score)
print("Neural Network Accuracy      :", net_score)
```
