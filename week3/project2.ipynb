```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "player":["Ronaldo","Messi","Mbappe","Vinijr","Neymar","Belingham","Nunomendes"],
    "Goals":[1000,900,500,200,400,200,100],
    "Assists":[400,500,200,100,300,100,100],
    "Matches":[1100,1000,500,500,600,400,300],
})
```

```python
df
```

```python
gpm = []

for i in range(len(df)):
    goal_per_match = df.loc[i, "Goals"] / df.loc[i, "Matches"]
    gpm.append(goal_per_match)

df["Goal_per_match"] = gpm

print(df)
```

```python
total = 0

for goal in df["Goals"]:
    total += goal

average_goal = total / len(df)
df["Average"] = average_goal
print(df)
print("Average Goals:", average_goal)
```

```python
top_scorer = df[df["Goals"] == df["Goals"].max()]
print(top_scorer)
```

```python
average= df["Goals"].mean()
above_average_goals = []
for value in df["Goals"]:
    if value >= average:
        above_average_goals.append("Top player")
    else:
        above_average_goals.append("Normal player")
df["Above_Average_goals"]=above_average_goals
print(df)
```
