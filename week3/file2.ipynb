```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame({
    "subject": ["math", "science", "english", "nepali"],
    "score": [98, 65, 76, 54]
})
```

```python
df
```

```python
for score in df["score"]:
    if score>=60:
        print("Pass")
    else:
        print("fail")
```

```python
df["score"].mean()
```

```python
df["score"].describe()
```

```python
mean_score = df["score"].mean()

df["mean_score"] = mean_score

print(df)
```

```python
mean_score = df["score"].mean()

df["above_average"] = df["score"] >= mean_score

for value in df["above_average"]:
    if value:
        print("True")
    else:
        print("False")
df
```

```python
True_data = df[df["above_average"]]
print(True_data)
```

```python
False_data= df[~df["above_average"]]
print(False_data)
```
