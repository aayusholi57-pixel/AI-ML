```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_val_score

# -----------------------------
# Load your dataset
# -----------------------------
# Replace with your dataset name if different
df = pd.read_csv("circles.csv")

# Features and target
X = df[["x", "y"]]
y = df["label"]

# -----------------------------
# 1. Logistic Regression on raw features
# -----------------------------
model = LogisticRegression()

score1 = cross_val_score(
    model,
    X,
    y,
    cv=5,
    scoring="accuracy"
).mean()

print(f"Accuracy using [x, y]: {score1:.2f}")

# -----------------------------
# 2. Engineer a new feature
# -----------------------------
df["dist2"] = df["x"]**2 + df["y"]**2

X_new = df[["x", "y", "dist2"]]

score2 = cross_val_score(
    model,
    X_new,
    y,
    cv=5,
    scoring="accuracy"
).mean()

print(f"Accuracy using [x, y, dist2]: {score2:.2f}")
```
