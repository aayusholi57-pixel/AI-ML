```python
import pandas as pd
import numpy as np

```

```python
class DataCleaner:

    def __init__(self, filename):
        self.filename = filename
        self.df = None

    def load_dataset(self):
        self.df = pd.read_csv(self.filename)
        return self

    def remove_duplicates(self):
        self.df.drop_duplicates(inplace=True)
        return self

    def fill_missing_mean(self, columns):
        for col in columns:
            self.df[col].fillna(self.df[col].mean(), inplace=True)
        return self

    def fill_missing_mode(self, columns):
        for col in columns:
            self.df[col].fillna(self.df[col].mode()[0], inplace=True)
        return self

    def detect_outliers(self, columns):

        for col in columns:

            Q1 = self.df[col].quantile(0.25)
            Q3 = self.df[col].quantile(0.75)

            IQR = Q3 - Q1

            lower = Q1 - 1.5 * IQR
            upper = Q3 + 1.5 * IQR

            self.df[col] = self.df[col].clip(lower, upper)

        return self

    def encode_categories(self, columns):

        self.df = pd.get_dummies(
            self.df,
            columns=columns,
            dtype=int
        )

        return self

    def standardize(self, columns):

        for col in columns:

            mean = self.df[col].mean()
            std = self.df[col].std()

            self.df[col] = (self.df[col] - mean) / std

        return self

    def save(self, output_file):

        self.df.to_csv(output_file, index=False)
        print("File saved successfully.")

        return self

    def summary(self):

        print("\nDataset Summary")
        print("-" * 40)
        print(self.df.info())
        print("\nMissing Values")
        print(self.df.isnull().sum())
        print("\nFirst 5 Rows")
        print(self.df.head())
```

```python
sample = pd.DataFrame({
    "age": [25, 30, np.nan, 40, 200, 25],
    "income": [50000, 60000, 55000, np.nan, 1000000, 50000],
    "city": ["NYC", "LA", "NYC", np.nan, "SF", "NYC"]
})

sample = pd.concat([sample, sample.iloc[[0]]], ignore_index=True)

sample.to_csv("sample_messy.csv", index=False)

cleaner = DataCleaner("sample_messy.csv")

(cleaner
 .load_dataset()
 .remove_duplicates()
 .fill_missing_mean(["age", "income"])
 .fill_missing_mode(["city"])
 .detect_outliers(["age", "income"])
 .encode_categories(["city"])
 .standardize(["age", "income"])
 .save("sample_cleaned.csv"))

cleaner.summary()

print("\nCleaned Dataset")
print(cleaner.df)
```
