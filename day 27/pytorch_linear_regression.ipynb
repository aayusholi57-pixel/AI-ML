```python
import torch

# Data
X = torch.linspace(-1, 1, 50)
Y = 2 * X + 1

# Things the model must learn
w = torch.tensor(0.0, requires_grad=True)
b = torch.tensor(0.0, requires_grad=True)

# How big each learning step is
lr = 0.1

for step in range(200):

    # 1. FORWARD
    pred = w * X + b

    # 2. LOSS
    loss = ((pred - Y) ** 2).mean()

    # 3. BACKWARD
    loss.backward()

    # 4. UPDATE
    with torch.no_grad():
        w -= lr * w.grad
        b -= lr * b.grad

    # 5. CLEAN OLD GRADIENTS
    w.grad.zero_()
    b.grad.zero_()

print("weight:", w.item())
print("bias:", b.item())
```
