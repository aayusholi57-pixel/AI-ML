```python
notes = [
    {
        "title": "Logistic Regression",
        "text": "Logistic regression is a classification algorithm used to predict categories such as positive or negative."
    },
    {
        "title": "Neural Networks",
        "text": "A neural network contains layers of connected neurons that learn patterns from training data."
    },
    {
        "title": "Overfitting",
        "text": "Overfitting happens when a model memorises the training data instead of learning patterns that generalise to new data."
    },
    {
        "title": "Cross Validation",
        "text": "Cross validation tests a model several times using different parts of the dataset and averages the results."
    },
    {
        "title": "Embeddings",
        "text": "Embeddings represent words or text as vectors of numbers so that similar meanings can be represented by nearby vectors."
    },
    {
        "title": "TF-IDF",
        "text": "TF-IDF converts text into numbers by giving importance to words based on how often they appear in a document and how rare they are across documents."
    },
    {
        "title": "Naive Bayes",
        "text": "Naive Bayes is a probabilistic machine learning algorithm that works especially well for many text classification problems."
    },
    {
        "title": "Early Stopping",
        "text": "Early stopping stops neural network training when validation performance stops improving, which can help prevent overfitting."
    },
]
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
texts = [note["text"] for note in notes]

vectorizer = TfidfVectorizer()

note_vectors = vectorizer.fit_transform(texts)
```

```python
def retrieve(query, k=3):

    # Convert the question into numbers
    query_vector = vectorizer.transform([query])

    # Compare question with all notes
    scores = cosine_similarity(
        query_vector,
        note_vectors
    )[0]

    # Get the highest scores
    best_indices = scores.argsort()[::-1][:k]

    results = []

    for i in best_indices:

        results.append(
            (
                float(scores[i]),
                notes[i]["title"],
                notes[i]["text"]
            )
        )

    return results
```

```python
results = retrieve(
    "what happens when a model memorises training data?",
    k=3
)

for score, title, text in results:

    print("\nScore:", round(score, 3))
    print("Title:", title)
    print("Text:", text)
```
